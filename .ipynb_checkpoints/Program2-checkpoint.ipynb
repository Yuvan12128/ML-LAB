```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier, export_text
# Step 1: Create the dataset
data = {
'Outlook': ['Sunny', 'Sunny', 'Overcast', 'Rain', 'Rain', 'Rain'],
'Temperature': ['Hot', 'Hot', 'Hot', 'Mild', 'Cool', 'Cool'],
'Humidity': ['High', 'High', 'High', 'High', 'Normal', 'Normal'],
'Wind': ['Weak', 'Strong', 'Weak', 'Weak', 'Weak', 'Strong'],
'PlayTennis': ['No', 'No', 'Yes', 'Yes', 'Yes', 'No'] }
df = pd.DataFrame(data)
print("The Data's in Table\n")
print(df,"\n")
# Step 2: Encode categorical data
le = LabelEncoder()
for column in df.columns:
    df[column] = le.fit_transform(df[column])
# Step 3: Split features and target
X = df.drop('PlayTennis', axis=1)
y = df['PlayTennis']
# Step 4: Train Decision Tree using Entropy (ID3)
model = DecisionTreeClassifier(
criterion='entropy',
random_state=0
)
model.fit(X, y)
# Step 5: Display the Decision Tree
print("Decision Tree:\n")
print(export_text(model, feature_names=list(X.columns)))
# Step 6: Classify a new sample
new_sample = pd.DataFrame({
'Outlook': ['Sunny'],
'Temperature': ['Cool'],
'Humidity': ['High'],
'Wind': ['Strong'] })
# Encode new sample using the same mapping
for column in new_sample.columns:
    encoder = LabelEncoder()
encoder.fit(data[column])
new_sample[column] = encoder.transform(new_sample[column])
prediction = model.predict(new_sample)
print("\nPrediction:")
if prediction[0] == 1:
    print("Play Tennis = Yes")
else:
    print("Play Tennis = No")
```
